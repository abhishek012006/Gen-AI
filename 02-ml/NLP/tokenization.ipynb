```python
corpus="""
hello my name is abhishek
i like to play football!
and do some coding...
"""
```

```python
print(corpus)
```

```python
import nltk

try:
    nltk.data.find("tokenizers/punkt_tab")
except LookupError:
    nltk.download("punkt_tab", quiet=True)
```

```python
##tokenization
##sentence-->paragraph
from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
```

```python
documents=sent_tokenize(corpus)
```

```python
for sentence in documents:
    print(sentence)
```

```python
##paragraph into words
##sentence into words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

```python
##Stemming
##classification problem
##comment of product is a positive review or negative review
#reviews--> eating, eat ,eated [going , gone, goes]
words=["eating","eats","eaten","writing", "writes","programming","final","finally"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
from nltk.stem import RegexpStemmer
reg_stemmer=RegexpStemmer("ing$|s$|able$", 3)
```

```python
reg_stemmer.stem("eating")
```

```python
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer("english")
```

```python
snowball_stemmer = SnowballStemmer("english")

for word in words:
    print(word+"---->"+snowball_stemmer.stem(word))
```
